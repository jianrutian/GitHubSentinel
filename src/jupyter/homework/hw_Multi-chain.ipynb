```python
# 试下moonshot模型的效果
from langchain_community.llms.moonshot import Moonshot
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
import os

# 初始化 ChatOpenAI 模型，指定使用的模型为 'gpt-4o-mini'
model = Moonshot(model="moonshot-v1-8k")

# 创建一个聊天提示模板，设置模板内容为"讲个关于 {topic} 的笑话吧"
prompt = ChatPromptTemplate.from_template("根据需求： {requirement}，写一段python代码")

# 初始化输出解析器，用于将模型的输出转换为字符串
output_parser = StrOutputParser()

# 构建一个处理链，先通过提示模板生成完整的输入，然后通过模型处理，最后解析输出
chain = prompt | model | output_parser

# 调用处理链，传入主题为"程序员"，生成关于程序员的笑话
print(chain.invoke({"requirement": "用动态规划解背包问题"}))
```

```python
# 传统langchain写法
from langchain import LLMChain, PromptTemplate
from langchain_community.llms.moonshot import Moonshot

# 定义 Python 代码生成的链
python_prompt = PromptTemplate(
    input_variables=["function_description"],
    template="""
    你是一个Python程序员。编写一个Python函数，执行以下任务:
    {function_description}
    """
)

python_chain = LLMChain(
    llm=Moonshot(model="moonshot-v1-8k"),
    prompt=python_prompt,
    output_key="python_code"
)

# 定义 Java 代码生成的链
java_prompt = PromptTemplate(
    input_variables=["function_description"],
    template="""
    你是一个Java程序员。编写一个执行以下任务的Java方法:
    {function_description}
    """
)

java_chain = LLMChain(
    llm=Moonshot(model="moonshot-v1-8k"),
    prompt=java_prompt,
    output_key="java_code"
)

from langchain.chains import SequentialChain

# 定义一个顺序链，先执行 Python 链，再执行 Java 链
multi_language_chain = SequentialChain(
    chains=[python_chain, java_chain],
    input_variables=["function_description"],
    output_variables=["python_code", "java_code"]
)

# 输入功能需求
function_description = "计算给定数的阶乘。"

# 运行链并获取代码
result = multi_language_chain.invoke(function_description)

# 输出结果
print("Python Code:")
print(result["python_code"])
print("\nJava Code:")
print(result["java_code"])
```

```python
# 本想用moonshot，但是我的账户有并发限制，后改成用chatgpt

from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_community.llms.moonshot import Moonshot
from langchain_core.runnables import RunnablePassthrough
from langchain_openai import ChatOpenAI

template = """
    你是一个python程序员。编写一段执行以下任务的python代码:
    {function_description}
    """
python_prompt = ChatPromptTemplate.from_template(template)
python_chain = (
    python_prompt
    | ChatOpenAI(model="gpt-4o-mini")
    | StrOutputParser()
)

template = """
    你是一个java程序员。编写一段执行以下任务的java代码:
    {function_description}
    """
java_prompt = ChatPromptTemplate.from_template(template)
java_chain = (
    java_prompt
    | ChatOpenAI(model="gpt-4o-mini")
    | StrOutputParser()
)

def output_result(result):
    for lang, code in result.items():
        print(f"\n{lang.capitalize()} Code:\n{code}")

coder_chain = (
     {"function_description": RunnablePassthrough()}
    |
    {
        "python": python_chain,
        "java": java_chain,
    } | RunnablePassthrough() | output_result
)
coder_chain.invoke("计算给定数的阶乘")
```
